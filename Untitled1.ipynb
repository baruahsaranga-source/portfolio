```python
# Import the requests library to handle HTTP requests
import requests

# Import the json library to format the output
import json

# Define the base URL for the Open-Meteo forecast API
url = "https://api.open-meteo.com/v1/forecast"

# Create a dictionary of parameters containing location coordinates and requested weather variables for 24 hours
params = {
    "latitude": 52.52,
    "longitude": 13.41,
    "hourly": "temperature_2m,relative_humidity_2m,wind_speed_10m",
    "forecast_hours": 24
}

# Send an HTTP GET request to the API with the specified URL and parameters
response = requests.get(url, params=params)

# Raise an exception if the HTTP request returned an error status code
response.raise_for_status()

# Parse the JSON response from the API into a Python dictionary
weather_data = response.json()

# Print the formatted JSON data to the console with an indentation of 2 spaces for readability
print(json.dumps(weather_data, indent=2))
```

```python
# Import the pandas library for data manipulation
import pandas as pd

# Extract the 'hourly' dictionary from the parsed JSON response
hourly_data = weather_data['hourly']

# Convert the dictionary into a pandas DataFrame (each key becomes a column, lists become rows)
df = pd.DataFrame(hourly_data)

# Rename the columns using a dictionary mapping old names to new names
df = df.rename(columns={
    'time': 'time',
    'temperature_2m': 'temperature_c',
    'relative_humidity_2m': 'humidity_pct',
    'wind_speed_10m': 'wind_speed_kmh'
})

# Optional: Convert the 'time' column from strings to actual datetime objects
df['time'] = pd.to_datetime(df['time'])

# Print the first few rows of the DataFrame to verify the output
print(df.head())
```

```python
# Calculate and print the number of missing values in the target columns before dropping
print("Missing values per column:\n", df[['temperature_c', 'wind_speed_kmh', 'humidity_pct']].isna().sum())

# Drop any rows containing missing values (NaN) specifically in the temperature, wind speed, or humidity columns
df = df.dropna(subset=['temperature_c', 'wind_speed_kmh', 'humidity_pct'])

# Reset the DataFrame index after dropping rows so the index remains a continuous sequence
df = df.reset_index(drop=True)

# Print a confirmation message showing the new shape of the DataFrame to verify rows were dropped
print(f"\nData cleaned. Current DataFrame shape: {df.shape}")
```

```python
# Print the current data types of all columns in the DataFrame to see what we are starting with
print("Original data types:\n", df.dtypes)

# Convert the temperature, wind speed, and humidity columns explicitly to float data types
df = df.astype({
    'temperature_c': float,
    'wind_speed_kmh': float,
    'humidity_pct': float
})

# Convert the 'time' column from strings to proper pandas datetime objects (if not already done)
df['time'] = pd.to_datetime(df['time'])

# Print the updated data types to verify that the conversions were successful
print("\nUpdated data types:\n", df.dtypes)
```

```python
# Check and print the number of duplicate rows based on the 'time' column
print(f"Duplicate rows found: {df.duplicated(subset=['time']).sum()}")

# Drop any duplicate rows based on the 'time' column, keeping the first occurrence
df = df.drop_duplicates(subset=['time'], keep='first')

# Reset the index to ensure it remains a continuous sequence after any rows are dropped
df = df.reset_index(drop=True)

# Print the final shape of the DataFrame to confirm its current size
print(f"Final DataFrame shape: {df.shape}")
```

```python
# Define a condition for impossible temperatures (below absolute zero: -273.15 C)
invalid_temp = df['temperature_c'] < -273.15

# Define a condition for impossible humidity (must be between 0% and 100%)
invalid_humidity = (df['humidity_pct'] < 0) | (df['humidity_pct'] > 100)

# Define a condition for impossible wind speeds (speed cannot be negative)
invalid_wind = df['wind_speed_kmh'] < 0

# Combine the conditions using the bitwise OR operator (|) to flag any row violating at least one rule
invalid_mask = invalid_temp | invalid_humidity | invalid_wind

# Filter the DataFrame using the mask to extract the rows with impossible values
invalid_rows = df[invalid_mask]

# Print the invalid rows so we can inspect what is being removed
print("Physically impossible rows detected:\n", invalid_rows)

# Filter the DataFrame to keep only the valid rows by using the bitwise NOT operator (~)
df = df[~invalid_mask]

# Reset the index to keep it continuous after potentially dropping rows
df = df.reset_index(drop=True)

# Print the final shape of the DataFrame to confirm the current size after cleaning
print(f"DataFrame shape after dropping impossible values: {df.shape}")
```

```python
# Import the gspread library's exceptions to handle cases where the sheet or tab doesn't exist
import gspread.exceptions

# Define the name of the main Google Spreadsheet
spreadsheet_name = 'weather'

# Try to open the spreadsheet to see if it already exists in your Google Drive
try:
    # Open the spreadsheet by its title
    sh = gc.open(spreadsheet_name)

    # Print a message confirming it was opened successfully
    print(f"Opened existing spreadsheet: '{spreadsheet_name}'")

# If the spreadsheet is not found, catch the specific exception and create a new one
except gspread.exceptions.SpreadsheetNotFound:
    # Create a new spreadsheet with the specified name
    sh = gc.create(spreadsheet_name)

    # Print a message confirming the creation of the new spreadsheet
    print(f"Created new spreadsheet: '{spreadsheet_name}'")

# Define the name of the specific worksheet (tab) we want to use
worksheet_title = 'Output'

# Try to access the specific worksheet to see if it already exists
try:
    # Get the worksheet by its title
    worksheet = sh.worksheet(worksheet_title)

    # Print a message confirming the worksheet was found (headers are intentionally skipped here)
    print(f"Worksheet '{worksheet_title}' already exists. Skipping header creation.")

# If the worksheet is not found, catch the exception, create it, and add the headers
except gspread.exceptions.WorksheetNotFound:
    # Add a new worksheet to the spreadsheet with the specified title, rows, and columns
    worksheet = sh.add_worksheet(title=worksheet_title, rows=1000, cols=10)

    # Print a message confirming the creation of the new worksheet
    print(f"Created new worksheet: '{worksheet_title}'")

    # Define the list of column headers to insert
    headers = ['time', 'temperature_c', 'wind_speed_kmh', 'humidity_pct']

    # Insert the headers as the first row in the newly created worksheet
    worksheet.insert_row(headers, index=1)

    # Print a message confirming the headers were written
    print("Headers written to the new worksheet.")
```
